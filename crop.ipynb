```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
#count all pixels in image
# Count ALL pixels in the image (total resolution)
def count_pixels(image):
    total_pixels = image.shape[0] * image.shape[1]  # Rows × Columns
    print("Total pixels in image:", total_pixels)
    return None
```

```python
def crop_img(normal, thermal):
    # plot
    plt.figure(figsize=(10, 5))
    normal_img = cv2.imread(normal)
    normal_img = cv2.cvtColor(normal_img, cv2.COLOR_BGR2RGB)
    thermal_img = cv2.imread(thermal)
    thermal_img = cv2.cvtColor(thermal_img, cv2.COLOR_BGR2RGB)
    plt.subplot(1, 2, 1)
    plt.imshow(normal_img)
    plt.axis('off')
    plt.title(normal.split("/")[-1])
    plt.subplot(1, 2, 2)
    plt.imshow(thermal_img)
    plt.axis('off')
    plt.title(thermal.split("/")[-1])
    plt.show()
    
    print("Normal image shape:", normal_img.shape)
    print("Thermal image shape:", thermal_img.shape)
    count_pixels(normal_img)
    count_pixels(thermal_img)

    #if horizontal image
    if normal_img.shape[0] < normal_img.shape[1]:
        x_start, y_start = 485, 360
        x_end, y_end = 2785, 2010
        
        cropped = normal_img[y_start:y_end, x_start:x_end]
        cropped = cv2.resize(cropped, (640, 480))

    #if vertical image
    if normal_img.shape[0] > normal_img.shape[1]:
        x_start, y_start = 430, 450
        x_end, y_end = 2140, 2700
                
        cropped = normal_img[y_start:y_end, x_start:x_end]
        cropped = cv2.resize(cropped, (480, 640))
    
    print("---------------------------------------")
    print("Normal image shape:", normal_img.shape)
    print("Thermal image shape:", thermal_img.shape)
    print("Resized normal image shape:", cropped.shape)
    count_pixels(normal_img)
    count_pixels(thermal_img)
    count_pixels(cropped)

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 3, 1)
    plt.imshow(normal_img)
    plt.axis('off')
    plt.title("Original Normal")
    plt.subplot(1, 3, 2)
    plt.imshow(cropped)
    plt.axis('off')
    plt.title("Cropped Normal")
    plt.subplot(1, 3, 3)
    plt.imshow(thermal_img)
    plt.axis('off')
    plt.title("Thermal")
    plt.show()

    folder_name = thermal.split("/")[-1]
    print(f"saving in  {folder_name}")
    #save the three images in a folder named thermal.split("/")[-1]
    if not os.path.exists(folder_name):
        os.makedirs(folder_name)
    normal_img_to_save = cv2.cvtColor(normal_img, cv2.COLOR_RGB2BGR)
    cropped_to_save = cv2.cvtColor(cropped, cv2.COLOR_RGB2BGR)
    thermal_img_to_save = cv2.cvtColor(thermal_img, cv2.COLOR_RGB2BGR)
    cv2.imwrite(f"{folder_name}/original_normal.jpg", normal_img_to_save)
    cv2.imwrite(f"{folder_name}/cropped_normal.jpg", cropped_to_save)
    cv2.imwrite(f"{folder_name}/thermal.jpg", thermal_img_to_save)
```
