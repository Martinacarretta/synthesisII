```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from PIL import Image
```

```python
#count all pixels in image
# Count ALL pixels in the image (total resolution)
def count_pixels(img):
    print("Image shape:", img.shape if isinstance(img, np.ndarray) else img.size)
    return None
```

```python
def crop_img(normal, thermal):
    # plot
    plt.figure(figsize=(10, 5))
    normal_img = Image.open(normal)
    thermal_img = Image.open(thermal)
    
    normal_img = np.array(normal_img)
    thermal_img = np.array(thermal_img)

    plt.subplot(1, 2, 1)
    plt.imshow(normal_img)
    plt.axis('off')
    plt.title(normal.split("/")[-1])
    plt.subplot(1, 2, 2)
    plt.imshow(thermal_img)
    plt.axis('off')
    plt.title(thermal.split("/")[-1])
    plt.show()
    
    print("Normal image shape:", normal_img.shape)
    print("Thermal image shape:", thermal_img.shape)
    count_pixels(normal_img)
    count_pixels(thermal_img)
    print("---------------------------------------")


    #if vertical image
    path_parts = normal.split("/")
    baby_ID = path_parts[7]
    if baby_ID == '29' or baby_ID =='30' or baby_ID =='31':
        print('using rotated crop of 29/30/31')
        normal_img = cv2.rotate(normal_img, cv2.ROTATE_90_COUNTERCLOCKWISE)
        
        x_start, y_start = 430, 450
        x_end, y_end = 2140, 2700
                
        cropped = normal_img[y_start:y_end, x_start:x_end]
        cropped = cv2.resize(cropped, (480, 640))
        cropped = cv2.rotate(cropped, cv2.ROTATE_90_CLOCKWISE)
        
    else:
        x_start, y_start = 485, 360
        x_end, y_end = 2785, 2010
        
        cropped = normal_img[y_start:y_end, x_start:x_end]
        cropped = Image.fromarray(cropped).resize((640, 480))
    
    print("---------------------------------------")
    print("Resized cropped image shape:", np.array(cropped).shape)
    count_pixels(cropped)

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 3, 1)
    plt.imshow(normal_img)
    plt.axis('off')
    plt.title("Original Normal")
    plt.subplot(1, 3, 2)
    plt.imshow(cropped)
    plt.axis('off')
    plt.title("Cropped Normal")
    plt.subplot(1, 3, 3)
    plt.imshow(thermal_img)
    plt.axis('off')
    plt.title("Thermal")
    plt.show()


    folder_name = thermal.split("/")[-1]
    print(f"saving in  {folder_name}")
    #save the three images in a folder named thermal.split("/")[-1]
    if not os.path.exists(folder_name):
        os.makedirs(folder_name)
    Image.fromarray(normal_img).save(f"{folder_name}/original_normal.jpg")
    Image.fromarray(cropped).save(f"{folder_name}/cropped_normal.jpg")
    Image.fromarray(thermal_img).save(f"{folder_name}/thermal.jpg")
```

```python
normal = "/home/martina/codi2/3year/synthesisII/crops/29/14.08.24-15.08.24/HM20240814213403_VIS.jpeg"
thermal = "/home/martina/codi2/3year/synthesisII/crops/29/14.08.24-15.08.24/HM20240814213403.jpeg"

crop_img(normal, thermal)
```
