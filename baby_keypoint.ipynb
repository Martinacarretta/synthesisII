```python
import cv2

# Load the thermal image
image_path = '/home/martina/codi2/3year/synthesisII/diferent/HM20240815055937.jpeg'
image = cv2.imread(image_path)
image_copy = image.copy()  # Use a copy to avoid modifying the original

# Define categories and colors for different keypoints
categories = {
    'h': 'hand',  
    'l': 'leg',  
    'a': 'arm',   
    't': 'torso'  
}
colors = {
    'hand': (0, 0, 255),   # Red
    'leg': (0, 255, 0),    # Green
    'arm': (255, 0, 0),    # Blue
    'torso': (0, 255, 255) # Yellow
}

keypoints = {cat: [] for cat in categories.values()}
current_category = 'hand'  # Default category

def click_event(event, x, y, flags, params):
    global image_copy

    if event == cv2.EVENT_LBUTTONDOWN:
        # Add the keypoint to the current category
        keypoints[current_category].append((x, y))

        # Draw the keypoint on the copied image
        cv2.circle(image_copy, (x, y), 5, colors[current_category], -1)

        # Display updated image with annotations
        cv2.imshow('Image', image_copy)

def key_event(key):
    global current_category

    key = chr(key).lower()  # Convert key to lowercase

    if key in categories:
        current_category = categories[key]
        print(f"Selected category: {current_category}")

    elif key == 'r':  # Press 'r' to reset the image
        global image_copy, keypoints
        image_copy = image.copy()
        keypoints = {cat: [] for cat in categories.values()}
        print("Reset keypoints.")
        cv2.imshow('Image', image_copy)

# Show image and set callback
cv2.imshow('Image', image_copy)
cv2.setMouseCallback('Image', click_event)

while True:
    key = cv2.waitKey(0) & 0xFF
    if key == 27:  # Press 'Esc' to exit
        break
    key_event(key)

cv2.destroyAllWindows()

# After the loop ends, redraw the final keypoints on the image
final_image = image.copy()

for category, points in keypoints.items():
    for (x, y) in points:
        cv2.circle(final_image, (x, y), 5, colors[category], -1)

# Display the final annotated image
cv2.imshow('Final Annotated Image', final_image)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Print keypoints at the end
print("Keypoints:", keypoints)
```
