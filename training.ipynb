```python
import torch
import cv2
import numpy as np
from pathlib import Path

# Load YOLOv5 model
def load_yolov5_model(weights_path):
    """
    Load the YOLOv5 model from the specified weights file.
    """
    model = torch.hub.load('ultralytics/yolov5', 'custom', path=weights_path, trust_repo=True)
    return model

# Detect babies in an image
def detect_babies(model, image_path, conf_threshold=0.1):
    """
    Detect babies in an image using the YOLOv5 model.
    """
    # Load image
    image = cv2.imread(image_path)
    if image is None:
        print(f"Error: Unable to load image at {image_path}")
        return None

    # Perform detection
    results = model(image)

    # Parse results
    detections = results.xyxy[0].numpy()  # Get detections in [x1, y1, x2, y2, confidence, class] format
    detected_image = image.copy()

    # Draw bounding boxes and labels
    for detection in detections:
        x1, y1, x2, y2, conf, cls = detection
        if conf >= conf_threshold:
            label = f"Baby {conf:.2f}"
            cv2.rectangle(detected_image, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)
            cv2.putText(detected_image, label, (int(x1), int(y1) - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

    return detected_image
```

```python
weights_path = "/home/martina/codi2/3year/synthesisII/yolov5s.pt"

# Path to the image you want to test
image_path = "/home/martina/codi2/3year/synthesisII/diferent/HM20240815055937.jpeg"  # Replace with the path to your image

# Load the YOLOv5 model
model = load_yolov5_model(weights_path)

# Detect babies in the image
detected_image = detect_babies(model, image_path)

if detected_image is not None:
    # Save or display the result
    output_path = "detected_image.jpg"
    cv2.imwrite(output_path, detected_image)
    print(f"Detection saved to {output_path}")

    # Display the result (optional)
    cv2.imshow("Detected Image", detected_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
