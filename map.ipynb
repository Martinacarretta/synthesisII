```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from PIL import Image
```

```python
def map_normal_to_cropped(baby_ID, x, y, normal_width, normal_height, normalized=True):
    if normalized:
        x = int(round(x * normal_width))
        y = int(round(y * normal_height))
        
    if baby_ID == '29' or baby_ID =='30' or baby_ID =='31':
        print('using rotated crop of 29/30/31')
        # Step 1: Rotate 90° counterclockwise
        x1 = y
        y1 = normal_width - 1 - x

        # Crop bounds (from rotated image)
        x_start, y_start = 430, 450
        x_end, y_end = 2140, 2700

        # Step 2: Check if point is in crop region
        if not (x_start <= x1 < x_end and y_start <= y1 < y_end):
            return None, None  # Point not in cropped region

        # Step 3: Adjust to cropped coordinates
        x2 = x1 - x_start
        y2 = y1 - y_start

        # Step 4: Resize to (480, 640)
        scale_x = 480 / (x_end - x_start)  # 480 / 1710
        scale_y = 640 / (y_end - y_start)  # 640 / 2250

        x3 = x2 * scale_x
        y3 = y2 * scale_y

        # Step 5: Rotate 90° clockwise
        x4 = int(round(639 - y3))
        y4 = int(round(x3))
        
    else:
        print("Using others")
        # Standard crop (no rotation), cropped to (640, 480)
        x_start, y_start = 485, 360
        x_end, y_end = 2785, 2010

        if not (x_start <= x < x_end and y_start <= y < y_end):
            return None, None  # Point not in cropped region

        # Translate to cropped image coordinates
        x_cropped = x - x_start
        y_cropped = y - y_start

        # Resize to (640, 480)
        scale_x = 640 / (x_end - x_start)  # 640 / 2300
        scale_y = 480 / (y_end - y_start)  # 480 / 1650

        x_resized = int(round(x_cropped * scale_x))
        y_resized = int(round(y_cropped * scale_y))

        x4 = x_resized
        y4 = y_resized

    return x4, y4
```

```python
def see (normal, thermal):
    path_parts = normal.split("/")
    baby_ID = path_parts[7]    
    # Load images
    normal_img = Image.open(normal)
    thermal_img = Image.open(thermal)

    # Convert to numpy arrays for plotting
    normal_np = np.array(normal_img)
    thermal_np = np.array(thermal_img)

    # Dimensions
    normal_width = normal_img.width
    normal_height = normal_img.height

    # Generate a random point within the bounds of the thermal image
    x = np.random.randint(0, normal_width)
    y = np.random.randint(0, normal_height)
    
    x2 = np.random.randint(0, normal_width)
    y2 = np.random.randint(0, normal_height)

    mapped_x, mapped_y = map_normal_to_cropped(baby_ID, x, y, normal_width, False)
    mapped_x2, mapped_y2 = map_normal_to_cropped(baby_ID, x2, y2, normal_width, False)

    print()
    print(f"Original Normal (x, y): ({x}, {y})")    
    print(f"Mapped to Cropped (x, y): ({mapped_x}, {mapped_y})")
    print(f"Original Normal (x, y): ({x2}, {y2})")    
    print(f"Mapped to Cropped (x, y): ({mapped_x2}, {mapped_y2})")

    # Plotting
    plt.figure(figsize=(12, 6))

    plt.subplot(1, 2, 1)
    plt.imshow(normal_np)
    plt.scatter(x, y, color='red', s=40)
    plt.scatter(x2, y2, color='red', s=40)

    plt.title(f"Original Normal\n(x, y) = ({x}, {y})")

    plt.subplot(1, 2, 2)
    plt.imshow(thermal_np)
    if mapped_x is not None:
        plt.scatter(mapped_x, mapped_y, color='white', s=40)
        plt.title(f"Mapped to Cropped\n(x, y) = ({mapped_x}, {mapped_y})")
    else:
        plt.title("Point outside crop area")
    if mapped_x2 is not None:
        plt.scatter(mapped_x2, mapped_y2, color='white', s=40)
        plt.title(f"Mapped to Cropped\n(x, y) = ({mapped_x2}, {mapped_y2})")
    else:
        plt.title("Point outside crop area")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```
