# YOLO pose detection model

While YOLO detects more images than Mediapipe, it is generally considered to be less accurate. YOLO, it uses fewer keypoints, which limits its ability to capture detailed pose information, especially for weird body positions. This reduced keypoint granularity can lead to less reliable or imprecise pose estimations, particularly in complex or occluded scenarios.

In contrast, Mediapipe provides higher-precision detections with a much richer set of 33 keypoints, covering not only the major joints but also minor landmarks like facial features, fingers, and toes. This makes it better suited for the task.

Moreover, Mediapipe includes additional metrics like visibility and presence scores for each keypoint, enabling more robust filtering and validation of the pose data. YOLO-based models typically lack this level of per-point metadata.

```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python
model = YOLO('yolov8n-pose.pt')  # Using the nano version for efficiency
```

```python
def test_pose_estimation(images_paths):
    num_images = len(images_paths)
    cols = 3  # Number of columns in the grid
    rows = (num_images // cols) + (num_images % cols > 0)  # Calculate rows for grid layout
    
    # Set up the figure for displaying the images
    plt.figure(figsize=(15, 5 * rows))
    
    for idx, image_path in enumerate(images_paths):
        # Load the image
        img = cv2.imread(image_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert to RGB for display
        
        if img.shape[1] > img.shape[0]:
            img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)  # Rotate by 90 degrees clockwise
            img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
            img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

        # Run inference
        results = model(img)
        
        # Get the image with detections
        img_with_detections = results[0].plot()
        
        # Display the result in the corresponding subplot
        plt.subplot(rows, cols, idx + 1)
        plt.imshow(img_with_detections)
        plt.axis('off')
        plt.title(f'Image {idx + 1}')
    
    plt.tight_layout()  # Adjust the layout for neat display
    plt.show()
```
